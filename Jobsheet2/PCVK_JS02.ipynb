```python
import cv2 as cv
from google.colab.patches import cv_imshow
from skimage import io
import matplotlib.pyplot as plt
import numpy as np

img = cv.imread('KTMlama.jpg')
cv_imshow(img)

print("Resolusi gambar: ", img.shape[1], " x ", img.shape[0])
```

```python
# Menampilkan gambar dengan format warna BGR
img = cv.imread('KTMlama.jpg')
plt.imshow(img)
```

```python
# Menampilkan gambar dengan format warna RGB
img = cv.imread('KTMlama.jpg')
img = cv.cvtColor(img, cv.COLOR_BGR2RGB)
plt.imshow(img)
```

```python
# Menampilkan citra grayscale

img_gray = cv.imread('KTMlama.jpg', cv.IMREAD_GRAYSCALE)

# tanpa cmap
plt.subplot(1,2,1)
plt.imshow(img_gray)
plt.title('Tanpa cmap')

# dengan cmap
plt.subplot(1,2,2)
plt.imshow(img_gray, cmap='gray')
plt.title('Dengan cmap')
```

```python
# Melakukan resizing

imgRGB = cv.cvtColor(img,cv.COLOR_BGR2RGB)
img_output = cv.resize(imgRGB, (347, 229))
plt.imshow(img_output)
```

```python
# Melakukan Flipping

imgFlip = cv.flip(img_output, 1)
plt.imshow(imgFlip)
```

```python
# Memperbesar ukuran canvas yang lebih besar dengan posisi gambar terbalik

imgFlip = cv.flip(img_output, 1)

fig = plt.figure(figsize=(10,10))
ax = fig.add_subplot(1,1,1)
ax.imshow(imgFlip)
```

```python
# Membentuk bentuk geometri 2D dari OpenCV. Diawali dengan pembuatan black image dengan tipe data int16

black_img = np.zeros(shape=(530,530,3), dtype=np.int16)
plt.imshow(black_img)
```

```python
# Menambah bentuk persegi panjang sesuai koordinat pt1 dan pt2

cv.rectangle(black_img, pt1=(290,0), pt2=(520,130), color=(0,0,255), thickness=10)
plt.imshow(black_img)
```

```python
# Menambahkan persegi panjang sesuai koordinat pt dan pt2 yang tertulis di kode program

cv.rectangle(black_img, pt1=(250,250), pt2=(350,350), color=(255,0,0), thickness=15)
plt.imshow(black_img)
```

```python
# Menambahkan lingkaran sesuai radius

cv.circle(black_img, center=(150,150), radius=60, color=(0,255,0), thickness=8)
plt.imshow(black_img)
```

```python
# Menambahkan lingkaran berwarna dan garis sesuai koordinat

cv.circle(black_img, center=(400,400), radius=50, color=(0,255,0), thickness=-1)
cv.line(black_img, pt1=(0,0), pt2=(530,530), color=(255,255,255), thickness=5)
plt.imshow(black_img)
```

```python
# Menambahkan Text

black_img = black_img.astype(np.uint8)
font = cv.FONT_HERSHEY_SIMPLEX
cv.putText(black_img, text = 'PCVK is fun!', org=(10,500), fontFace=font, fontScale=2,color=(255,255,0), thickness=3, lineType=cv.LINE_AA)
plt.imshow(black_img)
```

```python
# Membuat black image dengan tipe data int 8

black_img2 = np.zeros(shape=(530,530,3), dtype=np.int32)
plt.imshow(black_img2)
```

```python
# Meng-inisialisasi numpy array dengan tipe data int32

vertices = np.array([[150,350], [2250,250], [450,350], [250,450]], dtype=np.int32)
vertices
```

```python
# Mereshape array

pts = vertices.reshape((-1,1,2))
pts
```

```python
# Penambahan polyline

cv.polylines(black_img2, [pts], isClosed=True, color=(0,255,0), thickness=5)
plt.imshow(black_img2)
```

```python
# TUGAS
# Soal 1

# Membaca citra menggunakan OpenCV
img = cv.imread('KTMlama.jpg')

# Tampilkan dengan cv2_imshow (format BGR dieksekusi dengan benar)
print("--- Tampilan cv2_imshow ---")
cv_imshow(img)

# Tampilkan dengan plt.imshow (format BGR dianggap RGB)
print("--- Tampilan plt.imshow ---")
plt.imshow(img)
plt.title("Tanpa Konversi Warna (Tampak Biru/Keorenan)")
plt.show()

# Bukti nilai piksel pada koordinat (y=100, x=100)
pixel_val = img[100, 100]
print(f"Nilai piksel pada (100, 100) dalam BGR: {pixel_val}")
print(f"Penyebab: OpenCV menganggap BGR = [Blue: {pixel_val[0]}, Green: {pixel_val[1]}, Red: {pixel_val[2]}].")
print(f"Matplotlib menampilkan nilai BGR tersebut sebagai RGB = [Red: {pixel_val[0]}, Green: {pixel_val[1]}, Blue: {pixel_val[2]}].")
```
