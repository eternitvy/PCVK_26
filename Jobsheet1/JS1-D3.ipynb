```python
import matplotlib.pyplot as plt
import numpy as np

black_img = np.zeros(shape=(512,512,3), dtype=np.int16)
plt.imshow(black_img)

#Persegi panjang pada koordinat pt1 dan pt2
cv2.rectangle(black_img, pt1=(384, 8), pt2=(510, 150), color=(0, 255, 0), thickness=10)

#persegi panjang pada koordinat pt1 dan pt2 yang tertulis pada kode program
cv2.rectangle(black_img, pt1=(200, 200), pt2=(300, 300), color=(0, 0, 255), thickness=15)

#Lingkaran sesuai radius yang tertulis pada kode program
cv2.circle(black_img, center=(100, 100), radius=50, color=(255, 0, 0), thickness=8)

#Garis sesuai koordinat pt1 dan pt2
cv2.line(black_img, pt1=(0, 0), pt2=(512, 512), color=(255, 255, 255), thickness=5)

plt.imshow(black_img)
```

```python
font = cv2.FONT_HERSHEY_SIMPLEX
cv2.putText(black_img, text='Hello', org=(10, 500), fontFace=font, fontScale=4, color=(255, 255, 0), thickness=2, lineType=cv2.LINE_AA)

plt.imshow(black_img)
plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

# --- Inisialisasi img2 terlebih dahulu ---
# Baca gambar dan konversi ke RGB agar warna tepat
img_bgr = cv2.imread("/content/drive/MyDrive/AAPCVK/KTMlama.jpg")
img2 = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)

# ==========================================
# Soal 1
# ==========================================
imgRGBsmall = cv2.resize(img2, (400, 250))
plt.figure(figsize=(12, 6))
plt.imshow(imgRGBsmall)
plt.show()
print("Ukuran piksel matriks tetap:", imgRGBsmall.shape)

# ==========================================
# Soal 2
# ==========================================
# Kanal Red-Blue (Menolkan kanal Green)
img_RB = img2.copy()
img_RB[:, :, 1] = 0

# Kanal Green-Blue (Menolkan kanal Red)
img_GB = img2.copy()
img_GB[:, :, 0] = 0

fig, axes = plt.subplots(1, 2, figsize=(10, 5))
axes[0].imshow(img_RB)
axes[0].set_title("Channel Red-Blue")
axes[1].imshow(img_GB)
axes[1].set_title("Channel Green-Blue")
plt.show()

# ==========================================
# Soal 3
# ==========================================
crop1 = img2[20:115, 25:120]
plt.imshow(crop1)
plt.title("Crop Baris 20-115, Kolom 25-120")
plt.show()

# ==========================================
# Soal 4
# ==========================================
crop_red = img2[5:30, :, 0]  # Index 0 -> Channel Red
plt.imshow(crop_red, cmap='Reds')
plt.title("Baris 5-30, Semua Kolom, Red Channel")
plt.show()

# ==========================================
# Soal 5
# ==========================================
canvas = img2.copy()
h, w, _ = canvas.shape

for _ in range(5):
    pt1 = (np.random.randint(0, w-100), np.random.randint(0, h-100))
    pt2 = (pt1[0] + np.random.randint(30, 100), pt1[1] + np.random.randint(30, 100))
    color = (np.random.randint(0, 256), np.random.randint(0, 256), np.random.randint(0, 256))
    thickness = -1  # Fill solid rectangle
    cv2.rectangle(canvas, pt1, pt2, color, thickness)

plt.imshow(canvas)
plt.show()

# ==========================================
# Soal 6
# ==========================================
img_flipped = cv2.flip(img2, -1)  # -1 membalik vertikal sekaligus horizontal
plt.imshow(img_flipped)
plt.show()

# ==========================================
# Soal 7
# ==========================================
# Ganti path dengan lokasi foto aktivitas Anda
foto_diri = cv2.imread("/content/drive/MyDrive/AAPCVK/foto_aktivitas.jpeg")  # Sesuaikan path foto aktivitas
foto_diri_rgb = cv2.cvtColor(foto_diri, cv2.COLOR_BGR2RGB)

cv2.rectangle(foto_diri_rgb, (430, 350), (610, 580), (255, 0, 0), 3)
cv2.circle(foto_diri_rgb, (520, 465), 90, (0, 255, 0), 2)

plt.imshow(foto_diri_rgb)
plt.show()

# ==========================================
# Soal 8 dan 9
# ==========================================
img_Lenna = cv2.imread("/content/drive/MyDrive/AAPCVK/Lenna.png")

# Ambil channel Blue (OpenCV BGR -> Index 0)
blue_channel = img_Lenna[:, :, 0].copy()
blue_channel_rgb = cv2.cvtColor(blue_channel, cv2.COLOR_GRAY2RGB)

h, w = blue_channel.shape

# 8. Buat rectangle pada bagian sudut bawah kiri
cv2.rectangle(blue_channel_rgb, (10, h - 80), (110, h - 10), (255, 255, 255), 3)

# 9. Berikan tulisan nama file
font = cv2.FONT_HERSHEY_SIMPLEX
cv2.putText(blue_channel_rgb, 'Lenna.png', (15, h - 25), font, 0.5, (0, 255, 0), 2, cv2.LINE_AA)

plt.imshow(blue_channel_rgb)
plt.title("Blue Channel + Rectangle + Text")
plt.show()
```
